```
import pandas as pd
```

# Welcome to Colab!

```
df_train = pd.read_csv('/content/train.csv')
display(df_train.head())
```

## Data Cleaning and Preparation

```
# Display info about the DataFrame to check data types and non-null counts
display(df_train.info())
```

```
# Identify and handle missing values
# For simplicity, I'll drop rows with any missing values. You might choose a different strategy (e.g., imputation) based on the data.
initial_rows = df_train.shape[0]
df_train.dropna(inplace=True)
rows_after_missing_drop = df_train.shape[0]
print(f"Dropped {initial_rows - rows_after_missing_drop} rows with missing values.")

# Drop duplicate rows
initial_rows = df_train.shape[0]
df_train.drop_duplicates(inplace=True)
rows_after_duplicate_drop = df_train.shape[0]
print(f"Dropped {initial_rows - rows_after_duplicate_drop} duplicate rows.")
```

```
# Convert 'Order Date' and 'Ship Date' to datetime objects
df_train['Order Date'] = pd.to_datetime(df_train['Order Date'], format='%d/%m/%Y')
df_train['Ship Date'] = pd.to_datetime(df_train['Ship Date'], format='%d/%m/%Y')

# Display the DataFrame info again to confirm data type changes
display(df_train.info())
display(df_train.head())
```

```
# Export the cleaned data to a new CSV file
output_csv_path = '/content/train_cleaned.csv'
df_train.to_csv(output_csv_path, index=False)
print(f"Cleaned data exported to {output_csv_path}")
```
